```python
import pandas as pd
```

```python
claims_df = pd.read_csv('claims_and_billing.csv', sep=',', quotechar='"', engine='python', on_bad_lines='skip')
denials_df = pd.read_csv('denials.csv')
```

```python
# Merge claims with their respective denials
rca_df = pd.merge(claims_df, denials_df, on='claim_id', how='inner')
```

```python
# 1. Top Denial Reasons by VOLUME (Percentage)
print("Top 5 Denial Reasons (By Frequency - Percentage):")
percentage_denials = rca_df['denial_reason_description'].value_counts(normalize=True).head(5) * 100
print(percentage_denials.round(2))
```

```python
# 2. Top Denial Reasons by DOLLAR VALUE (Percentage)
print("\nTop 5 Denial Reasons (By Financial Impact - Percentage):")

total_denied_amount = rca_df['denied_amount'].sum()
financial_impact = rca_df.groupby('denial_reason_description')['denied_amount'].sum().sort_values(ascending=False).head(5)

percentage_financial_impact = (financial_impact / total_denied_amount) * 100
print(percentage_financial_impact.round(2))
```

```python
# 3. Insurance Provider with Highest Denial Amount
print("\nInsurance Provider with the Highest Denial Amount:")
highest_denial_provider = rca_df.groupby('insurance_provider')['denied_amount'].sum().idxmax()
highest_denial_amount = rca_df.groupby('insurance_provider')['denied_amount'].sum().max()
print(f"Provider: {highest_denial_provider}, Total Denied Amount: ${highest_denial_amount:,.2f}")
```
