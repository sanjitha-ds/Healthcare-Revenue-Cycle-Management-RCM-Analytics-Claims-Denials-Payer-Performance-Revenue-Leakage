```python
import pandas as pd
```

```python
claims = pd.read_csv('claims_and_billing.csv')
```

```python
#filter for only paid claims to evaluate contract yield
paid_claims = claims[claims['claim_status']=='Paid'].copy()
```

```python
#claculate reimbursement rate(yield) per payer
payer_yield = paid_claims.groupby('insurance_provider').agg(
    total_paid_claims=('claim_id','count'),
    total_billed=('billed_amount','sum'),
    total_paid=('paid_amount','sum')
).reset_index()
```

```python
payer_yield['contract_yield_pct']=(payer_yield['total_paid']/payer_yield['total_billed']*100)
print(payer_yield.sort_values(by='contract_yield_pct',
ascending=False).to_string(index=False))
```

```python
denials_by_payer = claims[claims['claim_status']=='Denied'].groupby(['insurance_provider',
'denial_reason']).size().reset_index(name ='count')
print(denials_by_payer.head(15))
```
