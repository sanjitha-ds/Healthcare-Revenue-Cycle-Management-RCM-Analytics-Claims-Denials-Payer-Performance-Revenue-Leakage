```python
import pandas as pd
import os
```

```python
print("Files:",os.listdir('.'))
claims=pd.read_csv('claims_and_billing.csv')
procedures=pd.read_csv('procedures.csv')
```

```python
print("claims cols:",claims.columns)
print("Procedure cols:",procedures.columns)
```

```python
#check for overlap
merged=procedures.merge(claims,on='encounter_id',how='left')
unbilled=merged[merged['claim_id'].isna()]
print("Total procedures:",len(procedures))
print("Unbilled procedures:",len(unbilled))
```
